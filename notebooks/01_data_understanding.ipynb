```python
q1 = df["Quantity"].quantile(0.25)
q3 = df["Quantity"].quantile(0.75)
iqr = q3 - q1                                  ##interquantile range
upper_limit = q3 + 1.5 * iqr                   ##important formula
outliers = df[df["Quantity"] > upper_limit]
print(upper_limit)
len(outliers)
```

```python
display(
    outliers[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID", "Country"]
    ].sort_values(by="Quantity", ascending=False).head(10)
)
```

```python
extreme_orders = df[df["Quantity"] >= 50000]
display(
    extreme_orders[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID", "Country"]
    ]
)
```

```python
extreme_invoice_details = df[df["InvoiceNo"].isin(extreme_orders["InvoiceNo"])]
display(
    extreme_invoice_details.sort_values(
        by=["InvoiceNo", "Quantity"],
        ascending=[True, False]
    )
)
```

```python
### High Quantity Outlier Investigation

The IQR method identified 56,635 records with quantities above the upper limit of 23.5 units.

Further investigation identified two extreme transactions with quantities exceeding 50,000 units:

- Invoice 541431: 74,215 units.
- Invoice 581483: 80,995 units.

Both transactions contain a single product line, have valid CustomerIDs, and have positive unit prices.

**Conclusion:** These records are potential extreme-value outliers. They should be flagged for further investigation rather than automatically removed, as they may represent genuine bulk purchases.
```

```python
lower_limit = q1-1.5*iqr
low_outliers = df[df["Quantity"] < lower_limit]
print(lower_limit)
```

```python
len(low_outliers)
```

```python
If an interviewer asks, "Why did you use the IQR method in your E-commerce analytics project?"

You can answer:
"I used the Interquartile Range method to identify potential outliers in the Quantity column. 
Since our retail dataset contains highly varying transaction quantities and some extreme values, 
IQR was suitable because it is less sensitive to extreme observations than mean-based methods. 
I calculated Q1 and Q3, obtained the IQR, and used the 1.5 multiplier to calculate the upper and lower boundaries. 
I identified 56,635 high-quantity records and investigated the extreme transactions separately instead of automatically removing them."
```

```python
display(
    low_outliers[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID", "Country"]
    ].sort_values(by="Quantity").head(15)
)
```

```python
low_outliers = low_outliers.copy()
low_outliers["Is_Cancellation"] = (
    low_outliers["InvoiceNo"].astype(str).str.startswith("C")
)
len(low_outliers)
```

```python
low_outliers["Is_Cancellation"].sum()
```

```python
~low_outliers["Is_Cancellation"].sum()
```

```python
### Low Quantity Outlier Investigation

The IQR method identified 1,984 records with quantities below the lower boundary of -12.5.

Further investigation classified these records into:

- Cancellation-marked invoices: 1,081 records (54.49%).
- Other invoice records: 903 records (45.51%).

The investigation also revealed descriptions indicating damaged products, discarded stock, incorrect stock entries and 
other inventory adjustments.

**Conclusion:** Negative quantity outliers represent different types of business transactions. 
They should not automatically be removed because some may contain important information about cancellations, 
returns and inventory adjustments.
```

```python
### Quantity Outlier Analysis – Final Findings

The Interquartile Range (IQR) method was used to identify potential outliers in the Quantity column.

**Calculated values:**
- Q1: 1
- Q3: 10
- IQR: 9
- Lower boundary: -12.5
- Upper boundary: 23.5

**Outlier findings:**
- High quantity outliers: 56,635
- Low quantity outliers: 1,984
- Total potential outliers: 58,619
- Percentage of potential outliers: 10.82%

Further investigation revealed that some extreme positive transactions had corresponding cancellation-marked records. 
Other negative quantity records included damaged stock, discarded products and inventory corrections.

**Conclusion:** The dataset contains a significant number of unusual quantity observations.
These records require business-context-based treatment rather than automatic removal.
```
